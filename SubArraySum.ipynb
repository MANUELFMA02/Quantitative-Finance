```python
def maxsumsubarray(array, k):
    currentsum = sum(array[:k])
    l = 0
    m = k
    for i in range(1, len(array) - 1):
        newsum = sum(array[i:i+k])
        if newsum > currentsum:
            l = i
        currentsum = max(currentsum, newsum)
        m = l+k-1
    return currentsum, l, m
```

```python
array = [1, 4, 0, 0, 1]
k = 2
suma, i, m = maxsumsubarray(array, k)
print(suma)
print(i)
print(m)
```

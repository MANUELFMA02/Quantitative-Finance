```python
def glasses(f, C, delta):
    t = 0
    v1 = 0
    v2 = 0
    v3 = 0
    while v3 < C:
        t += delta
        v1 = f*t
        if v1 > C:
            v2 = 0.5*(v1-C)
            if v2 > C:
                v3 = 0.5*(v2-C)
    return t            
            
```

```python
f = 1
C = 1
delta = 1
T = glasses(f,C,delta)
print(T)
```

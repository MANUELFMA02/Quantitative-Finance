```python
def counter(array):
    seen = {}
    for num in array:
        if num in seen:
            seen[num] += 1
        else:
            seen[num] = 1
    return seen       
```

```python
def maximum(array):
    counts = counter(array)
    n = len(array)
    for num, count in counts.items():
        if count > n // 2:
            return num
    return None
```

```python
array = [2, 3, 3, 3, 1]
k = maximum(array)
print(k)
```

```python
import numpy as np
           
```

```python
def numberofpaths(grid, moves):
    m, n = len(grid), len(grid[0])
    dp = [[0]*n for _ in range(m)]

    for i in range(m):
        for j in range(n):
            if grid[i][j] == 1:
                dp[i][j] = 0
                continue
            if i == 0 and j == 0:
                dp[i][j] = 1
                continue

            total = 0
            for dx, dy in moves:
                px, py = i - dx, j - dy
                if 0 <= px < m and 0 <= py < n:
                    total += dp[px][py]
            dp[i][j] = total

    return dp[m-1][n-1]     
```

```python
k = 8
l = 8
grid = [[0]*k for _ in range(l)]
moves = [(1,0), (0,1)]
paths = numberofpaths(grid, moves)
print("The total number of paths is = ", paths)
```

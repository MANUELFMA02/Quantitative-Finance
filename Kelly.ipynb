```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
outcomes = ["Heads", "Tails"]
n = 8
L = 1000
Total = np.zeros((L, n + 1))
Attempts = np.linspace(0, n, n + 1, endpoint=True)

p = 0.5
q = 1 - p
b = 2
weights = [p, q]
f = p - q / b

for k in range(L):
    capital = 100
    Total[k, 0] = capital
    for i in range(1, n + 1):
        Result = random.choices(outcomes, weights=weights, k=1)[0]
        if Result == "Heads":
            capital += b * f * capital
        else:
            capital -= f * capital
        Total[k, i] = capital
    plt.plot(Attempts, Total[k, :], marker='', color='b', linestyle='--', alpha = 0.005)

Mean = np.mean(Total, axis=0)
plt.plot(Attempts, Mean)
plt.title('Total Capital')
plt.xlabel('Attempts')
plt.ylabel('Capital $')
plt.show()
       
```
